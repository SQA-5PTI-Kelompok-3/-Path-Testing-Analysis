```python
# ==========================================
# PROGRAM NILAI AKHIR MAHASISWA UIN JAKARTA
# Path Testing Analysis
# ==========================================


print("=== PROGRAM NILAI AKHIR MAHASISWA ===")

# Input nilai mahasiswa
nilai_tugas = float(input("Masukkan nilai Tugas/Formative : "))
nilai_uts = float(input("Masukkan nilai UTS              : "))
nilai_uas = float(input("Masukkan nilai UAS              : "))


# Proses menghitung nilai akhir
nilai_akhir = (
    (nilai_tugas * 0.20) +
    (nilai_uts * 0.30) +
    (nilai_uas * 0.50)
)


# Menentukan nilai huruf
if nilai_akhir >= 80:
    grade = "A"

elif nilai_akhir >= 70:
    grade = "B"

elif nilai_akhir >= 60:
    grade = "C"

elif nilai_akhir >= 50:
    grade = "D"

else:
    grade = "E"


# Output hasil
print("\n=== HASIL PERHITUNGAN ===")
print("Nilai Tugas    :", nilai_tugas)
print("Nilai UTS      :", nilai_uts)
print("Nilai UAS      :", nilai_uas)

print("--------------------------")
print("Nilai Akhir    :", round(nilai_akhir,2))
print("Nilai Huruf    :", grade)
```
